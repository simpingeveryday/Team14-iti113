```python
%%writefile pipeline_src/recsys_score.py
import os
import glob
import boto3
import json
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

REGION = "ap-southeast-1"

s3_client = boto3.client('s3', region_name=REGION)
sm_runtime = boto3.client('sagemaker-runtime', region_name=REGION)

INPUT_DIR = "/opt/ml/processing/input"
OUTPUT_DIR = "/opt/ml/processing/output"

REGRESSION_ENDPOINT = "iti113-team14-airbnb-price-sls"
CLASSIFICATION_ENDPOINT = "iti113-team14-airbnb-instant-booking"

def get_price_predictions_in_batches(df, batch_size=1000):
    all_prices = []
    records = df.to_dict(orient='records')
    total_records = len(records)
    print(f"Starting price predictions for {total_records} records...")
    
    for i in range(0, total_records, batch_size):
        batch = records[i : i + batch_size]
        payload = json.dumps(batch)
        
        response = sm_runtime.invoke_endpoint(
            EndpointName=REGRESSION_ENDPOINT,
            ContentType="application/json",
            Accept="application/json",
            Body=payload
        )
        result = json.loads(response["Body"].read().decode("utf-8"))
        prices = [res["suggested_nightly_price"] for res in result]
        all_prices.extend(prices)
        
    return all_prices

def get_classification_predictions_in_batches(df, batch_size=1000):
    all_probs = []
    records = df.to_dict(orient='records')
    total_records = len(records)
    print(f"Starting classification predictions for {total_records} records...")
    
    for i in range(0, total_records, batch_size):
        batch = records[i : i + batch_size]
        payload = json.dumps({"data": batch})
        
        response = sm_runtime.invoke_endpoint(
            EndpointName=CLASSIFICATION_ENDPOINT,
            ContentType="application/json",
            Accept="application/json",
            Body=payload
        )
        result = json.loads(response["Body"].read().decode("utf-8"))
        probs = [res["probability_class_1"] for res in result]
        all_probs.extend(probs)
        
    return all_probs

def main():
    print("🚀 RecSys Processing Step Started!")
    
    csv_files = glob.glob(os.path.join(INPUT_DIR, "**", "*.csv"), recursive=True)
    if not csv_files:
        raise FileNotFoundError("No input CSV found in processing container input folder.")
    
    input_path = csv_files[0]
    print(f"Loading input file: {input_path}")
    
    raw_df = pd.read_csv(input_path, encoding="latin1", low_memory=False)
    
    # 1. Fallback for unknown synthetic cities
    VALID_CITIES = [
        "Paris", "New York", "Sydney", "Rome", "Rio de Janeiro", 
        "Istanbul", "Mexico City", "Bangkok", "Cape Town", "Hong Kong"
    ]
    if 'city' in raw_df.columns:
        raw_df.loc[~raw_df['city'].isin(VALID_CITIES), 'city'] = "Paris"

    # 2. Aggressively map booleans BEFORE inference for BOTH endpoints
    boolean_cols = ['host_is_superhost', 'host_has_profile_pic', 'host_identity_verified', 'instant_bookable']
    for col in boolean_cols:
        if col in raw_df.columns:
            raw_df[col] = raw_df[col].fillna('f')
            raw_df[col] = raw_df[col].astype(str).str.lower().str.strip().map(
                {'t': 1, 'true': 1, '1': 1, 'f': 0, 'false': 0, '0': 0}
            )
            raw_df[col] = raw_df[col].fillna(0).astype(int)

    # 3. Create strictly-cleaned dataframe for inference (replace NaN with None for JSON)
    df_for_inference = raw_df.astype(object).where(pd.notna(raw_df), None)

    print("Invoking Regression Endpoint...")
    raw_df["pred_price"] = get_price_predictions_in_batches(df_for_inference)
    
    print("Invoking Classification Endpoint...")
    raw_df["instant_book_prob"] = get_classification_predictions_in_batches(df_for_inference)
    
    print("Calculating Multi-Criteria Recommendation Scores...")
    
    # --- NEW: Safely clean the price column before doing math ---
    if 'price' in raw_df.columns:
        # Strip currency symbols and commas, convert to float
        raw_df['price'] = raw_df['price'].astype(str).str.replace(r'[\$,\s]', '', regex=True)
        raw_df['price'] = pd.to_numeric(raw_df['price'], errors='coerce')
        # Fill completely broken prices with a fallback median to prevent math crashes
        raw_df['price'] = raw_df['price'].fillna(raw_df['price'].median(skipna=True))
    # ------------------------------------------------------------
    
    # Calculate Residuals and Deals
    raw_df['price_residual'] = raw_df['price'] - raw_df['pred_price']
    raw_df['deal_value'] = raw_df['pred_price'] - raw_df['price']
    
    scaler = MinMaxScaler()
    raw_df['value_score'] = scaler.fit_transform(raw_df[['deal_value']])
    
    w_value, w_convenience = 0.5, 0.5
    raw_df['final_score'] = (w_value * raw_df['value_score']) + (w_convenience * raw_df['instant_book_prob'])

    print("Downloading historical master dataset...")
    BUCKET = "nyp-26s1-iti113"
    MASTER_KEY = "iti113/team14/data/airbnb-recommendation/processed/master_recsys_dataset.csv"
    
    try:
        master_obj = s3_client.get_object(Bucket=BUCKET, Key=MASTER_KEY)
        master_df = pd.read_csv(master_obj['Body'], low_memory=False)
        print(f"Loaded existing master list with {len(master_df)} records.")
        
        combined_df = pd.concat([master_df, raw_df], ignore_index=True)
        combined_df = combined_df.drop_duplicates(subset=['listing_id'], keep='last')
        print(f"Combined master list now has {len(combined_df)} records.")
        
    except s3_client.exceptions.NoSuchKey:
        print("No existing master list found in S3. Creating a new one.")
        combined_df = raw_df

    os.makedirs(OUTPUT_DIR, exist_ok=True)
    
    output_filename = os.path.basename(input_path).replace(".csv", "_scored.csv")
    raw_df.to_csv(os.path.join(OUTPUT_DIR, output_filename), index=False)
    combined_df.to_csv(os.path.join(OUTPUT_DIR, "master_recsys_dataset.csv"), index=False)
    
    print("✅ RecSys Scoring & Master Append Complete!")

if __name__ == "__main__":
    main()
```

```python
import sagemaker
from sagemaker.processing import ProcessingInput, ProcessingOutput
from sagemaker.sklearn.processing import SKLearnProcessor
from sagemaker.workflow.steps import ProcessingStep
from sagemaker.workflow.pipeline import Pipeline
from sagemaker.workflow.parameters import ParameterString
from sagemaker.workflow.pipeline_context import PipelineSession

# 1. Initialize Pipeline Session
pipeline_session = PipelineSession()

# 2. Define Parameter for incoming RecSys raw files

RECSYS_TRIGGER_PREFIX = f"iti113/{TEAM_ID}/trigger/input"

p_recsys_input = ParameterString(
    name="RecSysInputDataUrl", 
    default_value=f"s3://{BUCKET}/{PREFIX99}/recsys/raw/RecommendationSystemSyntheticListing.csv"
)

# 3. Initialize Scikit-learn Processor (reuses your team execution role)
recsys_processor = SKLearnProcessor(
    framework_version="1.2-1",
    instance_type="ml.m5.large",
    instance_count=1,
    role=role,
    sagemaker_session=pipeline_session,
    base_job_name=f"iti113-{TEAM_ID}-recsys-score",
    tags=[
        {"Key": "Course", "Value": "ITI113"},
        {"Key": "Semester", "Value": SEMESTER},
        {"Key": "Team", "Value": TEAM_ID},
        {"Key": "Student", "Value": STUDENT_ID},
    ],
)

# 4. Define the Processing Step
step_recsys_process = ProcessingStep(
    name="ScoreRecommendations",
    processor=recsys_processor,
    inputs=[
        ProcessingInput(
            source=p_recsys_input,
            destination="/opt/ml/processing/input",
        )
    ],
    outputs=[
        ProcessingOutput(
            output_name="scored",
            source="/opt/ml/processing/output",
            destination=f"s3://{BUCKET}/iti113/{TEAM_ID}/data/airbnb-recommendation/processed",
        )
    ],
    code="pipeline_src/recsys_score.py",
)

# 5. Assemble and Upsert the Pipeline
RECSYS_PIPELINE_NAME = f"iti113-{TEAM_ID}-recsys-scoring-trigger"

recsys_pipeline = Pipeline(
    name=RECSYS_PIPELINE_NAME,
    parameters=[p_recsys_input],
    steps=[step_recsys_process],
    sagemaker_session=pipeline_session,
)

recsys_pipeline.upsert(role_arn=role)
print(f"✅ Pipeline \"{RECSYS_PIPELINE_NAME}\" upserted successfully!")
```

```python
import time

def poll_pipeline_execution(execution_arn, sleep_seconds=30):
    """
    Poll a SageMaker Pipeline execution until it reaches a terminal status.
    """
    terminal_statuses = ["Succeeded", "Failed", "Stopped"]

    while True:
        desc = sm.describe_pipeline_execution(
            PipelineExecutionArn=execution_arn
        )

        status = desc["PipelineExecutionStatus"]

        print("=" * 80)
        print("Pipeline status:", status)
        print("Execution display name:", desc.get("PipelineExecutionDisplayName"))
        print("Start time:", desc.get("CreationTime") or desc.get("StartTime"))

        steps_response = sm.list_pipeline_execution_steps(
            PipelineExecutionArn=execution_arn
        )

        steps = steps_response.get("PipelineExecutionSteps", [])

        if steps:
            print("\nSteps:")
            for step in reversed(steps):
                step_name = step.get("StepName")
                step_status = step.get("StepStatus")
                failure_reason = step.get("FailureReason", "")

                print(f"- {step_name}: {step_status}")

                if failure_reason:
                    print(f"  Failure reason: {failure_reason}")
        else:
            print("\nNo steps listed yet.")

        if status in terminal_statuses:
            print()
            print("Final status:", status)
            return status

        print()
        print(f"Still running. Checking again in {sleep_seconds} seconds...")
        time.sleep(sleep_seconds)
```
