```python
import requests
import json

# 1. Define the API endpoint (This works because this notebook is on the same AWS machine)
url = "http://localhost:8000/recommend"

# 2. Define your search criteria (The JSON payload)
search_query = {
    "city": "Paris",
    "min_guests": 2,
    "max_price": 150.0,
    "w_value": 0.5,
    "w_convenience": 0.5
}

# 3. Send the POST request to the FastAPI server
print("Sending request to Recommendation API...")
response = requests.post(url, json=search_query)

# 4. Print the results!
if response.status_code == 200:
    data = response.json()
    print(f"✅ Success! Found {data['count']} listings.")
    print("-" * 40)
    
    # Print the top 3 recommendations beautifully
    for i, listing in enumerate(data['results'][:3]):
        print(f"Rank #{i+1}: Listing ID {listing['listing_id']}")
        print(f"   Price: €{listing['price']} (True Value: €{listing['pred_price']:.2f})")
        print(f"   Instant Book Prob: {listing['instant_book_prob']*100:.1f}%")
        print(f"   ML Score: {listing['ml_score']:.4f}")
        print("-" * 40)
else:
    print(f"❌ Error: {response.status_code}")
    print(response.text)
```
